```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import matplotlib.pyplot as plt

simulator = AerSimulator()
```

```python
# Part A: Correlated Outcomes

qc_a = QuantumCircuit(2, 2)

qc_a.h(0)
qc_a.cx(0, 1)

qc_a.measure([0, 1], [0, 1])

print("Part A - Correlated Outcomes")
display(qc_a.draw("mpl"))

result_a = simulator.run(qc_a, shots=100).result()
counts_a = result_a.get_counts()

print("Part A Results:", counts_a)
```

```python
# Part B: Opposite Outcomes

qc_b = QuantumCircuit(2, 2)

qc_b.h(0)
qc_b.cx(0, 1)
qc_b.x(1)

qc_b.measure([0, 1], [0, 1])

print("Part B - Opposite Outcomes")
display(qc_b.draw("mpl"))

result_b = simulator.run(qc_b, shots=100).result()
counts_b = result_b.get_counts()

print("Part B Results:", counts_b)
```

```python
# Part A Histogram
plt.figure(figsize=(6, 4))
plt.bar(counts_a.keys(), counts_a.values())
plt.xlabel("Measurement Outcomes")
plt.ylabel("Number of Shots")
plt.title("Part A - Correlated Outcomes")
plt.show()

# Part B Histogram
plt.figure(figsize=(6, 4))
plt.bar(counts_b.keys(), counts_b.values())
plt.xlabel("Measurement Outcomes")
plt.ylabel("Number of Shots")
plt.title("Part B - Opposite Outcomes")
plt.show()
```

```python
print("QUANTUM CORRELATION SWITCH - COMPARISON")
print("-" * 45)

print("\nPart A - Correlated Outcomes")
print("Prediction: 00 and 11")
print("Actual Results:", counts_a)

print("\nPart B - Opposite Outcomes")
print("Prediction: 01 and 10")
print("Actual Results:", counts_b)
```

### Challenge 4: Quantum Correlation Switch

#### Part A – Correlated Outcomes

**Prediction:** I predict that the dominant measurement outcomes will be 00 and 11, with approximately equal probabilities.

**Circuit:** I applied an H gate to qubit 0, followed by a CNOT gate with qubit 0 as the control and qubit 1 as the target. Then, I measured both qubits using 100 shots.

**Result:** The outcomes 00 and 11 were observed, with approximately equal counts.

#### Part B – Opposite Outcomes

**Prediction:** I predict that the dominant measurement outcomes will be 01 and 10, with approximately equal probabilities.

**Circuit:** I used the same circuit as Part A and added an X gate to qubit 1 before measurement. Then, I measured both qubits using 100 shots.

**Result:** The outcomes 01 and 10 were observed, with approximately equal counts.

#### Comparison

In Part A, the dominant outcomes were 00 and 11. Both qubits produced the same measurement value.

In Part B, the dominant outcomes were 01 and 10. The qubits produced opposite measurement values.

The difference between the circuits is the additional X gate applied to qubit 1 in Part B. This gate flips the second qubit's state, changing the correlated outcomes into opposite outcomes.

#### Conclusion

This experiment demonstrates that a small change in a quantum circuit can switch the relationship between two qubits' measurement outcomes. The H and CNOT gates create correlated outcomes, while adding an X gate to the second qubit changes the correlation into opposite outcomes. The experiment was performed using exactly two qubits and 100 shots for each circuit.
